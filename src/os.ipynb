### Check Existence

```julia
function path_existing_check(path)
    println("Current Path: $path")
    println( isfile(path) )
    println( isdir(path) )
    println( ispath(path) )
end
```

```julia
path_existing_check(origin_pwd)
```

```julia
path_existing_check("./os.ipynb")
```

```julia
path_existing_check("./os_ULTIMATE.ipynb")
```

### Fileinformation

filesize in bytes:

```julia
filesize("./os.ipynb")
```

```julia
function get_filesize(filepath)
    size_bytes = filesize(filepath)
    size_bits = size_bytes * 8
    size_kB = size_bytes / 1000
    size_mB = size_bytes / 1000000
    size_gB = size_bytes / 1000000000

    println("Filesize of '$filepath'")
    println("  - Bits: $size_bits")
    println("  - Bytes: $size_bytes")
    println("  - KB: $size_kB")
    println("  - MB: $size_mB")
    println("  - GB: $size_gB")
end

get_filesize("./os.ipynb")
get_filesize("../README.md")
get_filesize("../cover.png")
```

```julia
stat("./os.ipynb")
```

```julia
mtime("./os.ipynb")
```
